```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

```

```python
countriesoftheworld = pd.read_csv('countries-of-the-world.csv')

countriesoftheworld.head(10)
```

```python
gdp = countriesoftheworld['GDP ($ per capita)']

phones = countriesoftheworld['Phones (per 1000)']

# La columna usa coma como separador decimal (p. ej. "36,0"), así que pandas
# la lee como texto. Hay que convertirla a float o Seaborn la trata como categórica.
percent_literate = countriesoftheworld['Literacy (%)'].str.replace(',', '.').astype(float)

region = countriesoftheworld['Region']
```

Hacer un gráfico de dispersión con listas

En este ejercicio, utilizaremos un conjunto de datos que contiene información sobre 227 países. Este conjunto de datos contiene mucha información interesante sobre cada país, como las tasas de natalidad y mortalidad, y el producto interior bruto (PIB). El PIB es el valor de todos los bienes y servicios producidos en un año, expresado en dólares por persona.

Hemos creado tres listas de datos a partir de este conjunto de datos para que puedas empezar. gdp es una lista que contiene el valor del PIB por país, expresado en dólares por persona. phones es una lista del número de teléfonos móviles por cada 1000 personas en ese país. Por último, percent_literate es una lista que contiene el porcentaje de la población de cada país que sabe leer y escribir.

Crea un gráfico de dispersión del PIB (gdp) frente al número de teléfonos por cada 1000 personas (phones).

```python
sns.scatterplot(x=gdp, y=percent_literate)
plt.ylim(0, 100)  # Set the limits of the y-axis to be between 0 and 100
plt.show()
```

Hacer un gráfico de recuento con un DataFrame

En este ejercicio, examinaremos las respuestas a una encuesta enviada a los jóvenes. Nuestra pregunta principal aquí es: ¿cuántos jóvenes encuestados afirman tener miedo a las arañas? Se pidió a los participantes en la encuesta que estuvieran de acuerdo o en desacuerdo con la afirmación "Tengo miedo a las arañas". Las respuestas varían de 1 a 5, donde 1 es "Totalmente en desacuerdo" y 5 es "Totalmente de acuerdo".

Para empezar, la ruta del archivo csv con los datos de la encuesta se ha asignado a la variable csv_filepath.

Ten en cuenta que, como csv_filepath es una variable de Python, no necesitarás entrecomillarla cuando leas el csv.

Importa Matplotlib, pandas y Seaborn utilizando los nombres estándar.
Crea un DataFrame llamado df a partir del archivo csv situado en csv_filepath.
Utiliza la función countplot() con los argumentos x= y data= para crear un gráfico de recuento con los valores de la columna "Spiders" en el eje x.
Visualiza el gráfico.

```python
df = pd.read_csv('young-people-survey-responses.csv', index_col=0)   # Crea DataFrame a partir del archivo csv'

sns.countplot(x="Spiders", data=df)
plt.show()
```

¡Impresionante! Este gráfico nos muestra que la mayoría de los jóvenes informaron no tener miedo a las arañas.

Gráficos de tono y dispersión

En el vídeo anterior, aprendimos cómo hue nos permite hacer fácilmente subgrupos dentro de los gráficos Seaborn. Vamos a probarlo explorando los datos de los alumnos de secundaria. Tenemos mucha información sobre cada alumno, como su edad, dónde vive, sus hábitos de estudio y sus actividades extraescolares.

Por ahora, nos fijaremos en la relación entre el número de faltas que tienen en la escuela y su calificación final en el curso, segmentada por el lugar donde vive el alumno (zona rural frente a zona urbana).

Crea un gráfico de dispersión con "absences" en el eje x y la calificación final ("G3") en el eje y utilizando el DataFrame student_data. Colorea los puntos del gráfico en función de "location" (urbano vs. rural).

Haz que "Rural" aparezca antes que "Urban" en la leyenda del gráfico.

```python
student_data = pd.read_csv('student-alcohol-consumption.csv')

sns.scatterplot(x='absences', y='G3', hue='location', data=student_data, hue_order=['Rural', 'Urban'])
plt.legend(title="Location")
plt.title("Scatter plot of absences vs final grade by location")
plt.show()
```

Gráficos de tono y recuento

Sigamos explorando nuestro conjunto de datos de alumnos de secundaria examinando una nueva variable. La columna « "school" » (Instalaciones de la escuela) indica las iniciales de la escuela a la que asististe: «GP» o «MS».

En el último ejercicio, creamos un gráfico de dispersión en el que los puntos del gráfico se coloreaban en función de si el alumno vivía en una zona urbana o rural. ¿Cuántos alumnos viven en zonas urbanas frente a zonas rurales, y varía esto en función de la escuela a la que asiste el alumno? Hagamos un gráfico de recuento con subgrupos para averiguarlo.

Rellena el diccionario palette_colors para asignar el valor de ubicación "Rural" al color "green" y el valor de ubicación "Urban" al color "blue".

Crea un gráfico de recuento con "school" en el eje x utilizando el DataFrame student_data.

Añade subgrupos al gráfico utilizando la variable "location" y utiliza el diccionario palette_colors para que los subgrupos de ubicación sean verdes y azules.

```python
palette_colors = {"Rural": "green", "Urban": "blue"}

sns.countplot(x="school", hue="location", data=student_data, palette=palette_colors)

plt.show()
```
